## Statistics on Menascos Unknot with an Extra Stitch

```sagemath
V_extra_stitch=[(1,-1), (1,-11), (2,-2), (2,-9), (3,-3), (3,-16), (4,-4), (4,-17), (5,-6), (5,-15), (6,-8), (6,-14), (7,-1), (7,-12), (8,-2), (8,-19), (9,-9), (9,-18), (10,-11), (10,-13), (11,-17), (11,-21), (12,-16), (12,-23), (13,-6), (13,-10), (14,-8), (14,-12), (15,-10), (15,-13), (16,-21), (16,-25), (17,-15), (17,-20), (18,-3), (18,-5), (19,-23), (19,-26), (20,-4), (20,-7), (21,-14), (21,-22), (22,-20), (22,-24), (23,-22), (23,-25),(24,-5),(24,-18),(25,-24),(25,-26),(26,-7),(26,-19)];

G_extra_stitch=vertexset_to_XOGrid(V_extra_stitch)
#This function is from GridPyM and draws the grid diagram! 
draw_grid(G_extra_stitch, markings='XO')

Tries_extra_stitch=[]
#Below I unknotted ochiai unknot 1000 times and recorded how many steps the algorithm required each time
for n in range(1000):
    [simplified_G,tries] = simplify_grid(G_extra_stitch, effort='high',verbose=False)
    Tries_extra_stitch.append(tries)
```

```sagemath
#Trying to get an idea for the data
histogram(Tries_extra_stitch, bins=100)
```

```sagemath
#The histogram indicates an exponential distribution.  This seems to make sense to me.  I am going to best fit the data
import numpy as np
import scipy.stats as ss
import matplotlib.pyplot as plt
%matplotlib inline
P = ss.expon.fit(Tries_extra_stitch)
print(P)
```
