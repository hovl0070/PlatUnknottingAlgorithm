## The Original Culprit Unknot

```sagemath
#This is the inital vertex set of Culprit
V_culprit=[(1,-1),(1,-3),(2,-2),(2,-5),(3,-1),(3,-4),(4,-3),(4,-6),(5,-4),(5,-8),(6,-2),(6,-7),(7,-6),(7,-9),(8,-7),(8,-18),(9,-8),(9,-12),(10,-10),(10,-16),(11,-5),(11,-11),(12,-9),(12,-13),(13,-11),(13,-17),(14,-10),(14,-14),(15,-13),(15,-15),(16,-16),(16,-18),(17,-12),(17,-14),(18,-15),(18,-17)]

#The function below I made to convert vertex set to XO grid required for plotting with GridPyM
G_culprit=vertexset_to_XOGrid(V_culprit)

#This function is from GridPyM and draws the grid diagram! 
draw_grid(G_culprit, markings='XO')

#They have a very fast program that takes a grid diagram to a closed braid.
braid_culprit=convert_to_braid(G_culprit)
print('This link can be represented as the following braid word:\n\n',braid_culprit)
#Once you have the closed braid representation we can use snappy to compute knot invariants of the closed braid
B_culprit = ClosedBraid(braid_culprit)
B_culprit.simplify()
#print('This link has the following properties (knot floer homology calculations):\n\n',B.knot_floer_homology())

print('This link has Jones Polynomial:\n\n',B_culprit.jones_polynomial())  

#Once you have the closed braid representation we can use snappy to compute knot invariants of the closed braid
B_culprit.knot_floer_homology()
```

```sagemath

# Simplifies the given grid diagram, using an algorithm similar to Gridlink's built-in function (see http://homepages.math.uic.edu/~culler/gridlink/). There are two options: the "effort" variable can be equal to 'low', 'medium' or 'high'.
#These choices correspond to the number of random moves applied to the grid (not including destabilizations), which are 200, 1000 or 10000 respectively. If the "verbose" option is True (which is the default value) the amount of simplification is printed.

#I alternated the code so that it tells us how many steps/tries not counting destabilization it took to simplify, verbose must be set to True
[simplified_G_culprit,count] = simplify_grid(G_culprit, effort='high',verbose=True)
draw_grid(G_culprit)
draw_grid(simplified_G_culprit)
```
